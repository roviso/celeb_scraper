```python
def bio_summary(data):

    system_message = {
        "role": "system",
        # "content": prompts['system_message_content']
        "content": f'''You are a powerful ai tasked to summarize the biography of a celebrity with the data given to you in a predefined template as below:
        Keyword Bio: Early Life, Relationship, Career & Net Worth
Introduction
(The name of the person about whom the blog is written) stands out with her Hulk-like physique and muscular body in a world where beauty standards constantly evolve. She belongs to the ----- nationality and currently resides in ------.
(Person’s name) is a -------- (Profession)from ------ (Place) who has broken stereotypes and redefined being fit and strong.
This article discusses his/ her background, age, actual name, and Instagram presence, shedding light on her remarkable journey.
Keyword | Quick Facts

Real Name/Full Name

Nick Name

Gender

Known as

Date of Birth

Birthplace

Father

Mother

Siblings

Age

Height

Weight

Ethnicity

Nationality

Religion

Education

Profession

Monthly Income

Net Worth

Marital Status

Wife/Husband or (Bf/gf)

Marriage Date (if married)

Children... Fill in if the data in given.
        '''
    }
    
    user_message = {
        "role": "user", 
        "content": f'here is the data of celebrity i want to summarize: \n\n{data}'
    }

    messages = [
        system_message,
        user_message
    ]

    print(messages)

    MODEL = "gpt-3.5-turbo-0301"
    for i in range(1,4):
        # try:
        response = openai.ChatCompletion.create(
            model=MODEL,
            messages=messages,
        )
        print("openai respones: ", response)
        en_result = response.choices[0].message["content"]
        return en_result
```
